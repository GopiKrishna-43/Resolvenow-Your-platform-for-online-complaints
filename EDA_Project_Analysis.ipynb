```
df = pd.read_csv("/content/Student_performance.csv")
corr = df.corr(numeric_only=True)
plt.figure(figsize=(10, 6))
sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f", linewidths=0.5)
plt.title("Correlation Heatmap of Housing Dataset", fontsize=14)
plt.show()
```

```
sns.scatterplot(x="attendance_pct",y="math_score",hue="writing_score",data=df)
plt.show()
```

```
plt.hist(df["math_score"], bins = 10,color="lavender",edgecolor="pink")
plt.axvline(df["math_score"].mean(),color="blue",linestyle="dashed",linewidth=2)
plt.title("Histogram of marks")
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.show()
```

```
#barchart
avg_marks = df.groupby("extracurricular")["math_score"].mean()
print(avg_marks)
avg_marks.plot(kind='bar',color=['lavender','pink'])
plt.title("Average Marks by Gender")
plt.xlabel("extracurricular")
plt.ylabel("Average Marks")
plt.show()
```

```
import matplotlib.pyplot as plt
import pandas as pd
df = pd.read_csv("/content/Student_performance.csv")
plt.figure(figsize=(6, 5))
plt.boxplot([df["math_score"], df["reading_score"]], labels=["Math Score", "Reading Score"])
plt.title("Box Plot: Math vs Reading Scores")
plt.ylabel("Score")
plt.grid(True, linestyle="--", alpha=0.5)
plt.show()
```

attendance and study hours are positively related to academic performance (math, reading, and writing scores). Parental education levels vary among students, and participation in extracurricular activities does not seem to have a significant impact on average math scores in this dataset
